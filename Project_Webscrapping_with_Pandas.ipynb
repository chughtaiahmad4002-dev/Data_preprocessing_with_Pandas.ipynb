# 2) Read multiple files from multiple URLs with pandas and add a column

```python
import pandas as pd

# List of URLs containing tables
urls = [
    "https://en.wikipedia.org/wiki/List_of_continents_by_population",
    "https://en.wikipedia.org/wiki/List_of_countries_and_dependencies_by_population",
]

headers = {"User-Agent": "Mozilla/5.0"}

# 1. Create an empty list to store our DataFrames
dataframes = []

# 2. Use for loop to iterate through each URL
for url in urls:
    # Read all tables from the URL and grab the first one ([0])
    df = pd.read_html(url, storage_options=headers)[0]
    # Adding a new column
    df['Source'] = url
    # Append (add) this DataFrame to our list
    dataframes.append(df)
```

# 3) Organize all the Data in a Dictionary

```python
dict_world={
    "Continents":dataframes[0],
    "Countries":dataframes[1]
}
```
