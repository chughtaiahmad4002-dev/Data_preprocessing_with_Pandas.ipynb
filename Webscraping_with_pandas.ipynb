```python
import pandas as pd
import numpy as np
```

# 1) Read file from URL with **pandas**

```python
#Read file from the website
# 1) Scrape Continents Data from its specific URL
continents_url = "https://en.wikipedia.org/wiki/List_of_continents_by_population"
df_continents = pd.read_html(continents_url, storage_options={'User-Agent': 'Mozilla/5.0'})[0]

# 2) Scrape Countries Data from its specific URL
countries_url = "https://en.wikipedia.org/wiki/List_of_countries_and_dependencies_by_population"
df_countries = pd.read_html(countries_url, storage_options={'User-Agent': 'Mozilla/5.0'})[0]
```

# 2) Read multiple files from multiple URLs with pandas and add a column

```python
def run_scraping_pipeline(url):
    """
    Automated pipeline that extracts a Wikipedia table,
    cleans column names, and exports a ready-to-use CSV file.
    """
    print("Step 1: Extracting data from web...")
    dfs = pd.read_html(url, storage_options={'User-Agent': 'Mozilla/5.0'})
    df = dfs[0]

    print("Step 2: Cleaning and standardizing column names...")
    df = df.rename(columns={
        "% of world": "Percentage of World",
        "Source (official or from the United Nations)": "Reference (official or from the United Nations)"
    })

    df = df.dropna(how='all')

    print("Step 3: Exporting processed dataset...")
    output_filename = "cleaned_population_pipeline_output.csv"
    df.to_csv(output_filename, index=False)

    print(f"Pipeline complete! Saved file as: {output_filename}")
    return df

# --- Execute the Pipeline ---
target_url = "https://en.wikipedia.org/wiki/List_of_countries_and_dependencies_by_population"
pipeline_df = run_scraping_pipeline(target_url)
```
